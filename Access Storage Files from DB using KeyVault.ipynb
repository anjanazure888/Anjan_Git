### KEYVALUT-SECRET--SCOPE

- Created KeyVault from Azure Portal with the name AnjanKeyValut with valtAccessPolicy
- With in AnjanKeyVault, Created A SECRET  With AnjanSecret and value as 'accesskey' copied from ACCESSKEYs to access ADLS
- Create Scope using #secrets/createScope next to workspace URL
- Create Scope with Name as 'AnjanScope' , pass values DNS name = VaultURL; ResourceId=ResourceID from Keyvalut properties.

```python
storageaccount="anjanazure888dbstorage"
container="deletecontainer"
## SET the Value
accesskey=dbutils.secrets.get(scope="AnjanScope",key="AnjanSecret")
spark.conf.set("fs.azure.account.key."+storageaccount+".dfs.core.windows.net",accesskey)

path=f"abfss://{container}@{storageaccount}.dfs.core.windows.net/"
display(path)
display(dbutils.fs.ls(path))
```
