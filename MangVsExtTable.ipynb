```python
filepath="abfss://anjanazure888bronze@anjanazure888dbstorage.dfs.core.windows.net/circuits.csv"
df=spark.read.format("csv").option("header","true").load(filepath)
df.display()
```

```python
df.withColumnsRenamed({"lat":"latitude","lng":"longitude","alt":"altitude"})\
    .write.mode("overwrite")\
        .saveAsTable("circuits")
```

```python
%sql
DESCRIBE default.circuits
```

```python
%sql
DESCRIBE EXTENDED default.circuits
```
