### Using SAS Token Key

```python
storageaccount="anjanazure888dbstorage"
sastoken="GEt New SAS Token"

# Create any lingering account key --override first
# spark.conf.unset(f"fs.azure.account.key."+storageaccount+".dfs.core.windows.net")
# configure Spark to authenticate using container level SAS token
spark.conf.set(f"fs.azure.account.auth.type.{storageaccount}.dfs.core.windows.net","SAS")
spark.conf.set(f"fs.azure.sas.token.provider.type.{storageaccount}.dfs.core.windows.net","org.apache.hadoop.fs.azurebfs.sas.FixedSASTokenProvider")
spark.conf.set(f"fs.azure.sas.fixed.token.{storageaccount}.dfs.core.windows.net", sastoken)
display(dbutils.fs.ls("abfss://"+container+"@"+storageaccount+".dfs.core.windows.net/"))
```

```python
filename="circuits.csv"
container="anjanazure888bronze"
filepath=f"abfss://"+container+"@"+storageaccount+".dfs.core.windows.net/"+filename
display(filepath)
```

```python
df_SAS_circuit=spark.read.format("csv").option("header","true").load(filepath)
display(df_SAS_circuit)
```

### Access Connector

- Create Access Connector from Azure Portal (Access Connector for Azure Databricks)
        unity-catalog-access-connector is the default one which managed by DB alone
        create a fresh one for your purpose and copy resourceid to console
        /subscriptions/fdbb4778-7ded-4ecc-aa2b-1c5ef79cba55/resourceGroups/AnjanAzure888_ResourceGroup/providers/Microsoft.Databricks/accessConnectors/AnjanAccessConnector

- Access connector should have the storage blob data contributor role on Storage Account. For that you have to Create Role Assignment
        Go to Storage Account -- > Access Control (IAM) --> + ADD --> Add Role Assignment --> search storage blob data contributor and select --> NEXT --> Select Managed identity -- > select + members--> select created Accesscontroller

- Create Storage Credential & External location in DB workspace
        Go to DB workspace --> Catalog --> Catalog + --> create Credential -- >credential Type : azure managed identity ; credential name - anjancredential ; access connector Id - Resourceid that we copied ; --> create
        Go to Catalog --> create a external location --> ExternalLocationName : AnjanExtLoc ; storageType: AzureDatalakestorage; StorageCredential : Above created credential ; URL :abfss://anjanazure888bronze@anjanazure888dbstorage.dfs.core.windows.net/ --> Create

```python
filename="circuits.csv"
filepath=f"abfss://"+container+"@"+storageaccount+".dfs.core.windows.net/"+filename
container="anjanazure888bronze"
storageaccount="anjanazure888dbstorage"

dbutils.fs.ls(f"abfss://{container}@{storageaccount}.dfs.core.windows.net/")
```

```python
df=spark.read.format("csv").option("header","true").load(filepath)
df.display()
```
